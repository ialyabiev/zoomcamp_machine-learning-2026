```python
# 6) median horsepower
median1 = float(df['horsepower'].median())
print(f'median v1: {median1}')

most_frequent_horsepower = df["horsepower"].mode()[0]
print(f'most_frequent_horsepower= {most_frequent_horsepower}')

# fill na with most frequent
df["horsepower2"] = df["horsepower"]
df["horsepower2"] = df["horsepower2"].fillna(most_frequent_horsepower)
median2 = float(df['horsepower2'].median())
print(f'median v2: {median2}')
```

```python
# sum of weights
df2 = df[df['origin'] == 'Asia'][['vehicle_weight','model_year']]
X = df2.head(7).to_numpy()
XTX = X.T @ X

y = np.array([1100, 1300, 800, 900, 1000, 1100, 1200])

XTX_inv = np.linalg.inv(XTX)
w = XTX_inv @ X.T @ y
print('sum of weights =',format(w.sum(), ".3g"))
```
